### Identificando o numero de variaveis Object

```python
object_cols = df_carteira_cliente_traduzido.select_dtypes(include="object").columns.tolist() 
print('Total de variáveis do tipo Object:',len(object_cols))
print("---------------------------------------------")
print(object_cols)
```

```python
data_dict_object = pd.DataFrame({  # cria um DataFrame resumo (dicionário de dados) apenas para colunas do tipo object

    "variavel": object_cols,  # adiciona uma coluna com os nomes das variáveis do tipo object

    "Amostras (em ordem alfabética)": [", ".join(sorted(df_carteira_cliente_traduzido[col].dropna().astype(str).unique()))  # lista os valores únicos da coluna, ordenados alfabeticamente e separados por vírgula
                                       if df_carteira_cliente_traduzido[col].notna().any() else None  # se a coluna tiver apenas NaN, retorna None
                                       for col in object_cols],  # repete esse processo para cada coluna object

    "Amostras (em ordem de frequência)": [", ".join(df_carteira_cliente_traduzido[col].dropna().astype(str).value_counts().index.tolist())  # lista os valores únicos ordenados pela frequência (mais comum primeiro)
                                          if df_carteira_cliente_traduzido[col].notna().any() else None  # se só tiver NaN, retorna None
                                          for col in object_cols],  # repete para cada coluna object
    
    "qtd_valores_unicos": [df_carteira_cliente_traduzido[col].nunique()  # calcula a quantidade de valores únicos em cada coluna
                           for col in object_cols]

})

# Colocar em ordem alfabética
data_dict_object = data_dict_object.sort_values(
    by="Amostras (em ordem alfabética)",  # ordena o DataFrame com base nos valores das amostras em ordem alfabética
    ascending=True,  # ordem crescente
    na_position="last"  # coloca os valores None no final
).reset_index(drop=True)  # reseta o índice após a ordenação

data_dict_object  # exibe o DataFrame final com o resumo das colunas do tipo object
```

```python
for indice_linha in range(len(df_carteira_cliente_traduzido)):
    tmp_contrato = df_carteira_cliente_traduzido.loc[indice_linha, "tempo_de_contrato"]
    cbrc_total = df_carteira_cliente_traduzido.loc[indice_linha, "cobranca_total"]
    cbrc_mensal = df_carteira_cliente_traduzido.loc[indice_linha, "cobranca_mensal"]
    if pd.isnull(tmp_contrato) and pd.isna(cbrc_total):
        df_carteira_cliente_traduzido = df_carteira_cliente_traduzido.drop(indice_linha)
    else:
        df_carteira_cliente_traduzido.at[indice_linha, "tempo_de_contrato"] = round((cbrc_total / cbrc_mensal),0)
df_carteira_cliente_traduzido = df_carteira_cliente_traduzido.reset_index(drop=True)
```

```python
df_carteira_cliente_silver = df_carteira_cliente_traduzido.copy()

l_colunas = [
'servico_telefonico',
'multiplas_linhas', 
'servico_de_internet', 
'seguraca_online',
'backup_online', 
'protecao_de_dispositivos', 
'suporte_tecnico',
'streaming_de_tv', 
'streaming_de_filmes'
]
```

```python
for nome_coluna in df_carteira_cliente_silver.columns:
    if nome_coluna in l_colunas:
        for indice_linha in range(len(df_carteira_cliente_silver)):      
            registro_selecionado = df_carteira_cliente_silver.loc[indice_linha, nome_coluna].lower()
            if nome_coluna == "servico_de_internet":
                continue
            elif registro_selecionado != "no" and registro_selecionado != "yes":
                df_carteira_cliente_silver.at[indice_linha, nome_coluna] = "no"
```

```python
for indice_linha in range(len(df_carteira_cliente_silver)):
    registro_selecionado = df_carteira_cliente_silver.loc[indice_linha, "genero"]
    if registro_selecionado == "male":
        df_carteira_cliente_silver.at[indice_linha, "genero"] = "M"
    elif registro_selecionado == "female":
        df_carteira_cliente_silver.at[indice_linha, "genero"] = "F"
```
